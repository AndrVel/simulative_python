#### **Задача 1**

Ваша задача - выяснить сколько в среднем тратится времени на решение задачи.

**Примечание**: для правильного подсчета - рассчитайте сначала среднее время решения по каждой задаче в отдельности, и только затем находите общее среднее время решения задач.

Результат - число типа `float`, округлите до 2 знаков после запятой и запишите в переменную `res`.


**Решение**

Напишите свое решение ниже

```python
import csv
```

```python

# Ваше решение здесь
from datetime import datetime
from collections import defaultdict

rows = []
with open('codesubmit.csv', 'r', encoding='utf-8', newline='') as csv_file:
  # delimiter - разделитель в csv
  # DictReader создает словари
  reader = csv.DictReader(csv_file, delimiter=';')

  # надо по каждой задаче, по каждому ученику собрать список временных отметок.

  for row in reader:
    row['created_at'] = datetime.fromisoformat(row['created_at'])
    row['user_id'] = int(row['user_id'])
    row['problem_id'] = int(row['problem_id'])

    rows.append(row)
#  сортирую по времени события
rows.sort(key=lambda row: row['created_at'])

# Время первой активности пользователя по задаче
start_times = {}
# «Создай словарь problem_times. Если я обращусь к несуществующему ключу, автоматически создай для него пустой список».
problem_times = defaultdict(list)
# Уже учтённые решения
solved = set()


for row in rows:
  user_id = row['user_id']
  problem_id = row['problem_id']
  created_at = row['created_at']

  key = (user_id, problem_id)

  # Записываю в словарь если еще нет такого ключа
  # Запоминаем самое первое событие
  # if key not in start_times:
  #   start_times[key] = created_at
  start_times.setdefault(key, created_at)

  # Первая успешная отправка
  if (
      row['type'] == 'submit'
      and
      row['is_correct'] == '1'
      and
      key not in solved
  ):
    duration = created_at - start_times[key]
    seconds = duration.total_seconds()
    # Убираю гениальные решения
    if seconds > 0:
      problem_times[problem_id].append(seconds)
    solved.add(key)

# print(problem_times)

problem_averages = []
for problem_id, times in problem_times.items():
    average_time = sum(times) / len(times)
    problem_averages.append(average_time)


res = sum(problem_averages) / len(problem_averages)
res = round(res, 2)
print(res)
```

#### **Задача 2**

Ваша задача - выяснить сколько часов в среднем проводит юзер в день на платформе. Перерывы в активности за день - не учитываем.

Результат - число типа `float`, округлите до 2 знаков после запятой и запишите в переменную `res2`.

**Решение**

Напишите свое решение ниже

```python

# Ваше решение здесь

# По ключу (iesr_id, data) сделать список всех created_at
visits = defaultdict(list)

for row in rows:
  user_id = row['user_id']
  created_at = row['created_at']
  date = created_at.date()
  key = (user_id, date)

  # записываю в словарь
  visits[key].append(created_at)


# надо посчитать разницу в часах между последним и первым значением created_at для каждого ключа
for row in visits:
  start = min(visits[row])
  end = max(visits[row])
  duration = end - start
  seconds = duration.total_seconds()
  hours = seconds / 3600
  visits[row] = hours

res2 = sum(visits.values()) / len(visits)
res2 = round(res2, 2)


res2
```

#### **Задача 3**

Теперь давайте посмотрим на активные сеансы. Выясните, сколько задач в среднем решается за один активный сеанс.

**Активный сеанс** - период, когда между любой активностью пользователя разница менее или равна часу, не более

**Важно**: в расчет берем не только успешные попытки решений (`is_correct=1`), а и неуспешные тоже (`is_correct=0`), и тип `run` в том числе.

Результат - число типа `float`, округлите до 2 знаков после запятой и запишите в переменную `res3`.

**Решение**

Напишите свое решение ниже

```python
#Сортирую список по возрастанию даты/времени
rows.sort(key=lambda row: row['created_at'])

# делаю группировку по user_id
rows_groupped = defaultdict(list)
for row in rows:
  user_id = row['user_id']
  problem_id = row['problem_id']
  created_at = row['created_at']
  rows_groupped[user_id].append({'problem_id': problem_id,
                                 'created_at': created_at})
rows_groupped

# формирую активные сеансы более часа по каждому пользователю

user_sessions = defaultdict(list) # список с активностями пользователей

for id, activities in rows_groupped.items():
  # print(id, activities)
  problems_set = {activities[0]['problem_id']} # задаю начальное множество проблем, которые решал пользователь за сеанс

  session = [] # начальный список сессии
  # начальный момент
  previous_time = activities[0]['created_at']
  for n in range (len(activities)):
    # актуальное время
    actual_time = activities[n]['created_at']
    # проверка на длительность сеанса
    if (actual_time - previous_time).total_seconds() > 3600:
      # добавляю сформированный сет в sessions
      session.append(problems_set)
      problems_set = {activities[n]['problem_id']}
    else:
      # добавляю в сет очередную problem_id
      problems_set.add(activities[n]['problem_id'])
    previous_time = actual_time
  session.append(problems_set)
  user_sessions[id] = session

# user_sessions

# осталось посчитать сеансы каждого пользователя и количество задач, чтобы найти сколько задач за активный сеанс решает пользователь.
sessions_count = 0
problems_count = 0
for user_id, sessions in user_sessions.items():
  sessions_count += len(sessions)
  for session in sessions:
    problems_count += len(session)
res3 = round(problems_count / sessions_count, 2)
res3

```
