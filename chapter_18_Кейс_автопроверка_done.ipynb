```python
import inspect
import ast
import importlib

FORBIDDEN_NAMES = ['eval', 'exec']

def check_for_malicious_code(code, output_file):

  try:
    tree = ast.parse(code)
    for node in ast.walk(tree):
      # если тип node относится к ast.Attribute и node.attr входит в запрещенный список
      if isinstance(node, ast.Attribute) and node.attr in FORBIDDEN_NAMES:
        raise ValueError(f"Код содержит запрещенный атрибут: {node.attr}")
      elif isinstance(node, ast.Call) and isinstance(node.func, ast.Name) and node.func.id == "eval":
        raise ValueError(f"Код содержит вызов eval()")
  # если код содержит синтаксические ошибки
  except SyntaxError as e:
    with open(output_file, 'w') as f:
      f.write(f"Код содержит синтаксическую ошибку: {e.msg}\n")
    return str(e)
  # обработка исключения с вредоносным кодом
  except ValueError as e:
    with open(output_file, 'w') as f:
      f.write(f"Код содержит вредоносные элементы: {str(e)}\n")
      return str(e)
  with open(output_file, 'w') as f:
    f.write("Код не содержит вредоносные элементы.\n")

def run_tests(standard_code, user_code, test_cases, output_file):
  with open(output_file, 'w', encoding='utf-8') as f:
    for item in test_cases:
      # проверяю не является ли item просто числом.
      if isinstance(item, (int, float)):
        # это число, поэтому преобразовую в кортеж
        item = (item, )

      # Пытаюсь выполнить эталонный код
      try:
        # подаю в standard_code кортеж с распаковкой *
        standard_result = standard_code(*item)
      except Exception as e:
        f.write(f"Ошибка при выполнении эталонного кода: {e}\n")
        return
      # Теперь код пользователя
      try:
        users_result = user_code(*item)
      except Exception as e:
        f.write(f"Ошибка при выполнении кода пользователя: {e}")

      if standard_result == users_result:
        f.write('Тест пройден\n')
      else:
        f.write('Тест не пройден\n')
        f.write(f'Ожидаемый результат: {standard_result}\n')
        f.write(f'Результат пользователя: {users_result}\n')

# Чтение кодов из файлов
standard_module = importlib.import_module('etalon')
user_module = importlib.import_module('user_malicious')

test_cases = [(1, 2), (3, 4), (5, 6)]
output_file = 'output2.txt'

# Проверка на вредоносный код
malicious_code_message = check_for_malicious_code(inspect.getsource(user_module), output_file)

if malicious_code_message:
    with open(output_file, 'w') as f:
        f.write("Ошибка при проверке модуля user_malicious:\n")
        f.write(malicious_code_message)
else:
    run_tests(standard_module.add, user_module.add_numbers, test_cases, output_file)
```

```python
import multiprocessing
import time
import ast
import importlib
import inspect

FORBIDDEN_NAMES = ['eval']

def check_for_malicious_code(code, output_file):

  try:
    tree = ast.parse(code)
    for node in ast.walk(tree):
      # если тип node относится к ast.Attribute и node.attr входит в запрещенный список
      if isinstance(node, ast.Attribute) and node.attr in FORBIDDEN_NAMES:
        raise ValueError(f"Код содержит запрещенный атрибут: {node.attr}")
      elif isinstance(node, ast.Call) and isinstance(node.func, ast.Name) and node.func.id == "eval":
        raise ValueError(f"Код содержит вызов eval()")
  # если код содержит синтаксические ошибки
  except SyntaxError as e:
    with open(output_file, 'w') as f:
      f.write(f"Код содержит синтаксическую ошибку: {e.msg}\n")
    return str(e)
  # обработка исключения с вредоносным кодом
  except ValueError as e:
    with open(output_file, 'w') as f:
      f.write(f"Код содержит вредоносные элементы: {str(e)}\n")
      return str(e)
  with open(output_file, 'w') as f:
    f.write("Код не содержит вредоносные элементы.\n")

import multiprocessing

# это функция, которая сможет вернуть результат работы параллельного процесса
def wrapper(queue, func, *args):
    result = func(*args)
    queue.put(result)


def run_tests(standard_code, user_code, test_cases, output_file):

    # сюда можете вставить ваш код из прошлой задачи
    # в него необходимо будет добавить проверку на время выполнения


  with open(output_file, 'w', encoding='utf-8') as f:
    for item in test_cases:
      # проверяю не является ли item просто числом.
      if isinstance(item, (int, float)):
        # это число, поэтому преобразовую в кортеж
        item = (item, )

      # Пытаюсь выполнить эталонный код
      try:
        # подаю в standard_code кортеж с распаковкой *
        standard_result = standard_code(*item)
      except Exception as e:
        f.write(f"Ошибка при выполнении эталонного кода: {e}\n")
        return

      # Теперь код пользователя
      try:
        # это нужно, чтобы получить из параллельного процесса переменную user_result
        queue = multiprocessing.Queue()

        # измерение времени выполнения кода
        process = multiprocessing.Process(
            target=wrapper,
            args=(queue, user_code, *item)
        )
        # текущее время с начала эпохи 1970-01-01 00:00:00 UTC
        start_time = time.time()
        process.start() #запускаем процесс
        process.join(timeout=5) #ожидание процесса в течение 5 секунд

        if process.is_alive(): #проверка выполняется ли еще процесс
          process.terminate() #останавливаем процесс
          # вызываем исключение
          raise TimeoutError("Превышено время выполнения.")

        # получение переменной user_result
        user_result = queue.get()
        # расчет времени выполнения
        execution_time = time.time() - start_time
      # обработка вызванного исключения
      except TimeoutError as e:
        f.write("Предупреждение: время выполнения превышает 5 секунд.")
        return
      except Exception as e:
        f.write("Ошибка при выполнении кода пользователя: {}\n".format(e))
        return


      if standard_result == user_result:
        f.write('Тест пройден\n')
      else:
        f.write('Тест не пройден\n')
        f.write(f'Ожидаемый результат: {standard_result}\n')
        f.write(f'Результат пользователя: {user_result}\n')


# Чтение кодов из файлов
standard_module = importlib.import_module('etalon')
user_module = importlib.import_module('user_inf')

test_cases = [(1, 2), (3, 4), (5, 6)]
output_file = 'output3.txt'

# Проверка на вредоносный код


malicious_code_message = check_for_malicious_code(inspect.getsource(user_module), output_file)
if malicious_code_message:
    with open(output_file, 'w') as f:
        f.write("Ошибка при проверке модуля user_malicious:\n")
        f.write(malicious_code_message)
else:
    run_tests(standard_module.add, user_module.add_numbers, test_cases, output_file)


  # run_tests(standard_module, user_module, test_cases, output_file)
```
