#### **Задача 1**

Ваша задача написать функцию `count_success_and_failure`, которая принимает на вход путь к файлу с логами и подсчитывает количество успешных продлений и ошибок при списании. Функция должна вернуть кортеж из двух значений: количества успешных попыток и неуспешных.

**Решение**

Напишите свое решение ниже

```python

def count_success_and_failure(file_path):
  with open(file_path, 'r') as log:
    errors = 0
    success = 0
    for line in log:
      if line.split('|')[0].strip() == 'ERROR':
        errors += 1
        success -= 1
      elif line.split('|')[4].strip().startswith('[demon] Обновляем'):
        success += 1

  return success, errors

count_success_and_failure('auto_purchase.log')
```

#### **Задача 2**

Ваша задача написать функцию `auto_renewal_sub`, которая принимает на вход путь к файлу с логами и обрабатывает количество клиентов с автопродлением подписки. Мы хотим посмотреть на изменение этого показателя в динамике: посчитайте сглаженные значения с помощью метода скользящего среднего и метода медианного сглаживания.  

**Примечание:** При сглаживании берем все предыдущие значения, включая текущее, будущие значения не берем. Если в один день наблюдаем несколько записей об автопродлении - берем максимальное из имеющихся число клиентов с подпиской.

Функция должна записать в файл `auto_renewal_sub.txt` два списка, предварив их соответствущими обозначениями:

`Среднее: [2.0, 1.0, 0.67...]`

`Медиана: [2, 2, 0...]`

**Решение**

Напишите свое решение ниже

```python
# from datetime import datetime
def auto_renewal_sub(log_file_path):
  with open(log_file_path) as f:


    avg_list = []
    median_list = []

    res = {}

    # numbers = res.values()
    for row in f:
      # условие выбора строки с количеством клиентов
      condition = row.split('|')[-1].strip().startswith('[demon] Cегодня')
      if condition:
        # дата из строки
        date = row.split('|')[1].strip().split()[0]
        # количество из строки
        qty = int(row.split('|')[-1].split()[-1])
        counter = 1
        if (date not in res.keys()) or (res[date] < qty):
          res[date] = qty

          # нахожу очередное среднее
          intermediate_list = list(res.values())
          intermediate_average = round((sum(intermediate_list) / len(intermediate_list)), 2)
          avg_list.append(intermediate_average)

          # нахожу очередное медианное
          values = sorted(intermediate_list)
          n = len(intermediate_list)
          middle = n // 2
          if n % 2 == 1:
            median_list.append(values[middle])
          else:
            median_list.append(int((values[middle - 1] + values[middle]) / 2))

  # записываю результат в файл
  with open('auto_renewal_sub.txt', 'w', encoding='utf-8' ) as r:

    r.write('Среднее: [' + ', '.join(map(str, avg_list)) + ']\n')
    r.write('Медиана: [' + ', '.join(map(str, median_list)) + ']')

  return
```

```python

auto_renewal_sub('auto_purchase.log')
```

#### **Задача 3**

Напишите функцию `sub_renewal_by_day`, которая принимает на вход путь к файлу с логами и анализирует взаимосвязь дня продления подписки и количества продлений в этот день. Функция должна записать в файл `weekdays.txt` аналитическую записку в формате:

**`Количество обновлений подписки по дням недели:`**

**`Понедельник: 6`**

**`Вторник: 7`**

**`Среда: 8`**

**`...`**

**Решение**

Напишите свое решение ниже

```python
from datetime import datetime
```

```python
def sub_renewal_by_day(file_path):
  with open(file_path, 'r', encoding='utf-8') as f:

    report = {}

    for row in f:
      # условие, что подписка продлевается
      condition = row.split('|')[-1].strip().startswith('[demon] Обновляем')
      # условие, что подписка не продлилась из-за ошибки при списании
      condition_false = row.split('|')[-1].strip().find('ошибка при списании') > 0
      if condition:
        date = row.split('|')[1].split()[0]
        weekday = datetime.strptime(date, '%Y-%m-%d').weekday()
        if weekday in report:
          report[weekday] += 1
        else:
          report[weekday] = 1
      elif condition_false:
        report[weekday] -= 1
  # return report

  week = {
        0: 'Понедельник',
        1: 'Вторник',
        2: 'Среда',
        3: 'Четверг',
        4: 'Пятница',
        5: 'Суббота',
        6: 'Воскресенье'
    }

  with open('weekdays.txt', 'w', encoding='utf-8') as f1:
    # f1.write('Количество обновлений подписки по дням недели:\n')
    # for k, v in week.items():
    #   f1.write(f"{v}: {report[k]}\n")
    lines = ['Количество обновлений подписки по дням недели:']
    for k, v in week.items():
      lines.append(f'{v}: {report[k]}')
    f1.write('\n'.join(lines))

sub_renewal_by_day('auto_purchase.log')
```
