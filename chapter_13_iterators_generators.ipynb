# Python ДЗ 9: Генератор колоды карт

Напишите генераторную функцию card_deck(), которая будет возвращать все возможные комбинации из двух параметров: номинала карты и масти (пик, бубей, червей, крестей).  

Важно: всего в колоде 52 карты, разделенные на четыре масти по 13 карт каждая.

```python
def card_deck():
    suits = ['Пик', 'Бубей', 'Червей', 'Крестей']
    ranks = [*range(2, 11), 'J', 'Q', 'K', 'A']
    for suit in suits:
        for rank in ranks:
            yield f'{rank} {suit}'
```

```python
card_deck()
```

```python
a = card_deck()
```

```python
next(a)
```

Список достоинств карт:

range(2, 11) даёт числа 2…10 (11 не включается).

Оператор распаковки * разворачивает их прямо внутрь списка.

Потом добавляются строки 'J', 'Q', 'K', 'A' (валет, дама, король, туз).

```python
[*range(2, 11), 'J', 'Q', 'K', 'A']
```

Если нужно увидеть сразу все карты, то можно написать так:

```python
cards = list(card_deck())
```

# Python ДЗ 9: Извлечение букв из строки

Напишите функцию show_let(s), которая принимает в качестве аргумента строку s и использует генераторное выражение для извлечения всех букв из строки.

Функция должна возвращать каждую найденную букву по одной, начиная с первой.

Если в строке нет букв, функция должна вернуть ошибку StopIteration.

```python
def show_let(s):
  for letter in s:
    if letter.isalpha():
      yield letter
```

```python
# res = show_let('Python is a popular programming language')
res = show_let('a!SD')
```

```python
print(next(res))
```

Реализуйте функцию substring_partitions, которая принимает 2 аргумента - строку string и число N.

Функция должна возвращать все возможные разбиения строки string на подстроки длиной не более N.

```python
def substring_partitions(string, N):
    # база рекурсии: пустая строка -> одно разбиение: пустой кортеж
    if not string:
        yield ()
        return

    # перебираем длину первого куска: от 1 до N,
    # но не дальше конца строки
    for first_len in range(1, min(N, len(string)) + 1):
        first = string[:first_len]      # первый кусок
        rest = string[first_len:]       # остаток строки

        # рекурсивно получаем все разбиения остатка
        for tail in substring_partitions(rest, N):
            # добавляем наш первый кусок в начало
            yield (first,) + tail
```

Нужно перебрать все способы порезать строку на куски длиной ≤ N, сохранив порядок символов. То есть это задача на рекурсию / перебор, а не на split().

Возьмём твой пример: string = 'abcd', N = 3.

Что вообще считается «разбиением»
Разбиение — это выбор всех возможных подстрок подряд, которые вместе снова дают 'abcd', и каждая подстрока не длиннее 3.

Например:

'a' | 'b' | 'c' | 'd'

'a' | 'b' | 'cd'

'a' | 'bc' | 'd'

'a' | 'bcd'

'ab' | 'c' | 'd'

'ab' | 'cd'

'abc' | 'd'

Каждый вариант — кортеж подстрок.

Как к этому прийти (идейно)
Подход рекурсией «берём первый кусок, к остатку применяем ту же функцию»:

На каждом шаге мы выбираем первую подстроку — это любой префикс строки длиной от 1 до N.
Для 'abcd' и N = 3 возможные первые куски:

'a' (осталось 'bcd')

'ab' (осталось 'cd')

'abc' (осталось 'd')

Для «остатка» строки делаем то же самое — вызываем функцию рекурсивно и получаем все варианты его разбиений.

К каждому варианту разбиения остатка просто приписываем выбранный первый кусок в начало.

База рекурсии: если строка пустая, есть ровно одно «разбиение» — пустой кортеж (ничего не осталось резать).

```python
list(substring_partitions('abcd', 3))
```

Если переданная лямбда-функция равна None- нужно выполнить сложение по умолчанию (сложение чисел).

Функция должна возвращать итератор, который будет складывать соответствующие элементы списков согласно переданной лямбда-функции либо по умолчанию.

Важно: проверка будет проводится вызовом метода next - убедитесь, что это работает с вашей функцией.

```python
list1 = [1, 2, 3, 4]
list2 = [10, 20, 30, 40, 50]
list3 = [100, 200, 300, 400, 500]

add_lists = lambda a, b, c: a * b + c
```

```python
from itertools import zip_longest
def sum_of_lists(list1=[], list2=[], list3=[], func=(lambda a, b, c: a+b+c)):
  for i, j, k in zip_longest(list1, list2, list3):
    a = i if i else 0
    b = j if j else 0
    c = k if k else 0
    if func:
      yield func(a, b, c)
    else:
      yield a + b + c
```

## Эталонное решение и разбор

```python
def sum_of_lists(*lists, func=None):
    if func is None:
        func = lambda *args: sum(args)
    return iter(map(lambda x: func(*x), zip(*lists)))
```

*lists — позиционные аргументы, упакованные в кортеж.
То есть можно вызвать так: sum_of_lists([1, 2, 3], [10, 20, 30])

Внутри функции lists будет равно ([1, 2, 3], [10, 20, 30]).

func=None — опциональный параметр, функция, которую нужно применить к «столбцам» элементов. Если её не передали, будет использована сумма.

Значение по умолчанию для func

    if func is None:
        func = lambda *args: sum(args)

Если func не передана, её делают анонимной функцией

lambda *args: sum(args).

*args в lambda означает, что она принимает любое количество аргументов и возвращает их сумму.

То есть по умолчанию «функция для столбца» — просто сумма элементов этого столбца.



zip(*lists) — группировка по позициям

zip берёт элементы с одинаковым индексом из нескольких последовательностей.

*lists распаковывает кортеж lists в отдельные аргументы функции zip.

Например:

a = [1, 2, 3]

b = [10, 20, 30]

list(zip(a, b))  # [(1, 10), (2, 20), (3, 30)]

Если ты вызовешь:

sum_of_lists([1, 2, 3], [10, 20, 30])
то внутри:


zip(*lists)  # -> zip([1,2,3], [10,20,30]) -> итератор по (1,10), (2,20), (3,30)

Каждый элемент x в map ниже будет, например, (1, 10), потом (2, 20) и т.д.

map(lambda x: func(x), zip(lists))

Разобьём:

zip(*lists) даёт кортежи по позициям: (1, 10), (2, 20), (3, 30) и т.д.


map(...) применяет функцию к каждому такому кортежу.


Внутренняя функция:

lambda x: func(*x)

x — это кортеж, например (1, 10).

func(*x) вызывает func(1, 10) за счёт распаковки *x.

Если func — сумма по умолчанию, то:

для (1, 10) → sum(1, 10) → 11

для (2, 20) → 22

для (3, 30) → 33

Таким образом map(...) даёт итератор по результатам func для каждого «столбца».

iter(...) вокруг map(...) тут на самом деле лишний: map и так уже возвращает итератор, так что можно было просто return map(...). Но это не ломает работу.
