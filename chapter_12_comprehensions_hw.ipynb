# Python ДЗ 8: Наибольший однозначный делитель

Для всех чисел от 1 до 1000 найдите наибольшую однозначную цифру, на которую делится каждое из чисел.

Пояснение: Делители числа 1000: 1, 2, 4, 5, 8, 10, 20, 25, 40, 50, 100, 125, 200, 250, 500, 1000. Максимальный делитель, являющийся однозначной цифрой - 8.

Напишите функцию highest_divisible_digit, которая возвращала бы словарь вида {число: наибольший однозначный делитель, ...}.

```python
def highest_divisible_digit():
  return {num1: num2 for num1 in range(1, 1001) for num2 in range(1,10) if num1%num2 == 0}
```

# Python ДЗ 8: Удаление по индексу

Напишите функцию remove_num, которая принимает список чисел lst и возвращает этот список без элементов с индексами 0, 2, 4 и 6.

```python
def remove_num(lst):
  return [x for x in lst if lst.index(x) not in (0, 2, 4,6)]
```

```python
lst = [12,24,35,70,88,120,155]
remove_num(lst)
```

```python
# Эталонное решение
def remove_num(lst):
  return [value for index,value in enumerate(lst) if index not in [0,2,4,6]]
```

# Python ДЗ 8: Шифрование строк

Напишите функцию encryption, которая принимает на вход строку и зашифровывает её.

Зашифровка происходит следующим образом: строчные буквы от а до x должны быть заменены на следующую букву по алфавиту, а буквы y и z должны остаться неизменными. заглавные буквы не шифруются.

```python
def encryption(string):
  alphabet = list('abcdefghijklmnopqrstuvwxyz')
  return ''.join([alphabet[alphabet.index(x) + 1] if (x in alphabet) and (x not in ('y', 'z')) and x.islower() else x for x in string])
```

```python
encryption('The lazy cat sat on the windowsill and watched the birds outside')
```

# Python. ДЗ 9: Сумма с условием

На вход функции sum_with_condition подается массив arr, состоящий из произвольного количества чисел.

Задача заключается в том, чтобы посчитать сумму всех элементов, которые без остатка делятся на 3.

Важно: Задачу нужно решить именно с помощью list comprehensions - постарайтесь избегать использования циклов.

```python
arr = [1, -1, 0, -3, 3, 9, 15, 15.1]
```

```python
def sum_with_condition(arr):
  return sum([x for x in arr if x%3 == 0])
```

```python
sum_with_condition(arr)
```

# Python. ДЗ 8: Кортеж из списков

На вход функции flat_tuple подается кортеж tup, состоящий из произвольного количества списков. Эти списки, в свою очередь, состоят из произвольных элементов - это могут быть и числа, и строки, и None, и другие коллекции.

Задача состоит в том, чтобы максимально простым способом сделать такой кортеж «плоским». Другими словами, нужно вернуть список, состоящий из элементов всех подсписков исходного кортежа.

```python
tup = ([1], [2, '2'], [[1, 2], [3, 4]], [], ['s'])
```

```python
def flat_tuple(tup):
  return [y for x in tup for y in x]
```

На вход функции array_cond_processing подается массив arr, состоящий из произвольного количества элементов произвольного типа.

Задача заключается в том, чтобы вернуть новый массив, трансформированный по следующим правилам:

Если i элемент старого массива - список, целое число или кортеж,
  то вернуть True
  Иначе вернуть False
Важно: Задачу нужно решить именно с помощью list comprehensions - постарайтесь избегать использования циклов.

```python
def array_cond_processing(arr):
  return [True if isinstance(x, (int, list, tuple)) else False for x in arr]
```

```python
arr = [1, 2, (1), 's', [1, 2], None, {1: 2}, (5, )]
```

# Python ДЗ 8: Уникальные слова

Напишите функцию get_unique_words(), которая на вход принимает список строк sentences - предложения для анализа, и множество stopwords - список стоп-слов, которые нужно исключить из результата.

Функция должна вернуть список уникальных слов, которые не являются стоп-словами и содержатся в предложениях.

Примечание: в результате все буквы должны быть строчными. Результирующий список должен быть отсортирован в алфавитном порядке.

```python
def get_unique_words(sentences, stopwords):
  res = [y.lower() for x in sentences for y in x.split() if y.lower() not in stopwords]
  res.sort()
  return res
```

```python
sentences = ["Python is a popular programming language", "It was created by Guido van Rossum in 1991"]
stopwords = {'a', 'by', 'in', 'is', 'it', 'of', 'the', 'to', 'was', '1991', 'it', 'programming', 'popular', 'language', 'it'}
```

```python
get_unique_words(sentences, stopwords)
```

# Python Дз 8: Конвертация букв

Напишите функцию convert_to_numbers, которая принимает на вход строку s и заменяет все буквы в строке на соответствующие числа, где 1 соответствует 'a', 2 - 'b' и так далее до 26 - 'z'.

Регистр букв не должен иметь значения. Если символ не является буквой алфавита, он будет заменен на пробел.

```python
def convert_to_numbers(s):
  alphabet = 'abcdefghijklmnopqrstuvwxyz'
  return [alphabet.index(x.lower()) + 1 if x.lower() in alphabet else ' ' for x in s]
```

```python
alphabet = 'abcdefghijklmnopqrstuvwxyz'
```

```python
a = 'Hello World!'
```

```python
convert_to_numbers(a)
```

эталонное решение

```python
import string

def convert_to_numbers(s):
  dict25 = {a: i+1 for a, i in zip(string.ascii_lowercase, range(26))}
  return [dict25.get(a.lower(), ' ') for a in s]
```

# Python. ДЗ 5: Найти максимальное значение меньше заданного [2]

Условие задачи

Кортеж nums содержит произвольное количество целых чисел в любом порядке. limit - содержит число, ограничивающее максимум искомого значения. Необходимо найти наибольшее значение строго меньше заданного лимита. Если такого значения нет, вернуть -1.

Задание

Напишите функцию lim_max, которая возвращает ограниченный максимум.

Функция lim_max принимает на вход nums - исходный кортеж чисел и limit - числовое ограничение сверху.

Примечание: Для тренировки работы с условиями, воспользуйтесь однострочным val = a if True else b.

Примечание: Решите задачу с помощью генерации списка (list comprehensions).

```python
nums = (10, 20, 30, 40, 50, 60, 70, 80, 100)
limit = 50
```

```python
def lim_max(nums, limit):
  res = max([x for x in nums if x<limit]) if [x for x in nums if x<limit] else -1
  return res
```

```python
lim_max(nums, limit)
```
