# Python ДЗ 7: Сортировка сотрудников по зарплате

```python
employees = [ {'name': 'Иванов', 'position': 'секретарь', 'salary': 50000},
                         {'name': 'Сидоров', 'position': 'менеджер', 'salary': 45000},
                         {'name': 'Петров', 'position': 'директор', 'salary': 100000} ]
```

```python
def sort_employees(employees):
  # возвращает новый список, отсортированный по возрастанию зарплаты и содержащий только должности менеджер и директор.
  filtered = list(filter(lambda x: x['position'] in ('менеджер', 'директор'), employees))
  return sorted(filtered, key=lambda x: x['salary'])
```

```python
sort_employees(employees)
```

# Python ДЗ 7: Сумма квадратов четных чисел

принимает список чисел и возвращает сумму квадратов только четных чисел в этом списке

```python
from functools import reduce
def sum_squared_even(lst):
  # выбираю четные
  evens = list(filter(lambda x: x%2 == 0, lst))
  # возвожу в квадрат
  squares = list(map(lambda x: x**2, evens))
  # нахожу сумму. чтобы пустой список не ломал код надо указывать начальное значение 0 третьих параметром.
  return reduce(lambda x, y: x + y, squares, 0)
```

```python
sum_squared_even([1, 2, 3, 4, 5])
```

```python
sum_squared_even([0])
```

# Python ДЗ 7: Задача о студентах

найти студентов, у которых балл выше среднего значения, отсортировать их по имени и вернуть список только имен студентов.

```python
students = [("Алиса", 20, 4.0), ("Марк", 21, 3.5), ("Иван", 19, 3.8), ("Мария", 22, 3.2)]
```

```python
def calculate_top_students(students):
  # список средних баллов
  means = list(map(lambda x: x[2], students))
  # средний балл
  mean_mark = sum(means) / len(means)
  # отбираем с баллом выше среднего
  above_mean = list(filter(lambda x: x[2] > mean_mark, students))
  # список имен
  names = list(map(lambda x: x[0], above_mean))
  # сортирую
  names.sort()
  return names
```

```python
calculate_top_students(students)
```

# Python. ДЗ 11: Лямбда-функции и reduce

Необходимо реализовать функцию find_max, которая принимает на вход список arr. Список arr имеет произвольную длину.

Функция должна возвращать максимальное значение в этом списке. Подразумевается, что список arr состоит только из чисел.

Важно: Задачу обязательно нужно решить с помощью функции reduce и лямбда-функции.

Примечание: Если на вход подан пустой список, необходимо вернуть None.

```python
arr = [1, 20, -3, 5, 0]
```

```python
def find_max(arr):
    # ваше решение ниже
    if arr:
        res = reduce(lambda x, y: x if x > y else y, arr)
    else:
        res = None
    return res
```

# Python. ДЗ 7: Сортировка словаря

Необходимо вернуть словарь, отсортированный по заданным правилам. Строка type - тип сортировки: по ключам (keywise) или по значениям (valuewise). Строка order - порядок сортировки: по убыванию desc или по возрастанию asc.

```python
def sort_dict(d, type, order):
  sort_way = 0 if type == 'keywise' else 1
  sort_direction = True if order == 'desc' else False
  res = dict(sorted(d.items(), key=lambda x: x[sort_way], reverse=sort_direction))
  return res
```

```python
d = {'a': 1, 'b': 5, 'c': 0}
type = 'keywise'
order = 'desc'
```

# Python. ДЗ 6: Прибавить число к каждому элементу списка

Дан список чисел ref_list. Необходимо к каждому его элементу с позиции start до stop включительно, прибавить значение num. Если ref_list короче или равен start или stop вернуть -1. Значение start может быть больше, меньше или равно stop.

Напишите функцию list_add_num, которая возвращает:

исходный список, в котором числа на позициях от start до stop увеличены на num;
если список ref_list короче или равен start или stop вернуть -1
Функция list_add_num принимает на вход

исходный список ref_list
целые числа позиций start и stop
целое число num
Примечание: Решите задачу без использования цикла.

```python
def list_add_num(ref_list, start, stop, num):
  if start >= len(ref_list) or stop >= len(ref_list):
    return -1
  else:
    if start <= stop:
      return list(map(lambda x: x+num if start <= ref_list.index(x) <= stop else x, ref_list))
    else:
      return list(map(lambda x: x+num if stop <= ref_list.index(x) <= start else x, ref_list))
```
